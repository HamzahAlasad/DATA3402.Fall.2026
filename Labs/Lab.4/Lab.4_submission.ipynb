6. Implement a triangle class analogous to the rectangle and circle in question 5.

```python
import math

class Triangle(ShapeCalculations):
    def __init__(self, a=0.0, b=0.0, c=0.0, x_cord=0.0, y_cord=0.0):
        super().__init__(x_cord, y_cord)
        self.__a = float(a)
        self.__b = float(b)
        self.__c = float(c)

    # getters
    def a(self):
        return self.__a

    def b(self):
        return self.__b

    def c(self):
        return self.__c

    # setters
    def set_a(self, a):
        self.__a = float(a)

    def set_b(self, b):
        self.__b = float(b)

    def set_c(self, c):
        self.__c = float(c)

    def compute_perimeter(self):
        return self.__a + self.__b + self.__c

    def compute_area(self):
        # Semi-perimeter for Heron's Formula
        s = self.compute_perimeter() / 2
        area = s * (s - self.__a) * (s - self.__b) * (s - self.__c)
        return math.sqrt(area)


My_Triangle = Triangle()

print("--Triangle--")
My_Triangle.set_a(3)
My_Triangle.set_b(4)
My_Triangle.set_c(5)
My_Triangle.set_x_cord(0)
My_Triangle.set_y_cord(2)
print(f"side a: {My_Triangle.a()}")
print(f"side b: {My_Triangle.b()}")
print(f"side c: {My_Triangle.c()}")
print(f"xcord: {My_Triangle.x_cord()}")
print(f"ycord: {My_Triangle.y_cord()}")
print(f"area: {My_Triangle.compute_area()}")
print(f"perimeter: {My_Triangle.compute_perimeter()}")
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the perimeter of the object. 

```python
import math

class ShapeCalculations:
    def __init__(self, x_cord=0, y_cord=0):
        self.__x_cord=x_cord
        self.__y_cord=y_cord

    # getters
    def x_cord(self):
        return self.__x_cord

    def y_cord(self):
        return self.__y_cord

    # setters
    def set_x_cord(self, x_cord):
        self.__x_cord=x_cord

    def set_y_cord(self, y_cord):
        self.__y_cord=y_cord

    def compute_area(self):
        raise NotImplementedError

    def compute_perimeter(self):
        raise NotImplementedError
        
    def get_perimeter_points(self):
        raise NotImplementedError


class Rectangle(ShapeCalculations):
    def __init__(self, length=0, width=0, x_cord=0, y_cord=0):
        super().__init__(x_cord, y_cord)
        self.__length=length
        self.__width=width

    # getters
    def length(self):
        return self.__length

    def width(self):
        return self.__width

    # setters
    def set_length(self, length):
        self.__length=length

    def set_width(self, width):
        self.__width=width

    def compute_area(self):
        return self.__length * self.__width

    def compute_perimeter(self):
        return (self.__length + self.__width) * 2
        
    def get_perimeter_points(self):
        pts = []
        x, y = self.x_cord(), self.y_cord()
        w, l = self.width(), self.length()
        
        # 4 points per side to make 16 points
        for i in range(4): pts.append((x + w * (i/4), y)) # bottom edge
        for i in range(4): pts.append((x + w, y + l * (i/4))) # right edge
        for i in range(4): pts.append((x + w * (1 - i/4), y + l)) # top edge
        for i in range(4): pts.append((x, y + l * (1 - i/4))) # left edge
        return pts


class Circle(ShapeCalculations):
    def __init__(self, radius=0, x_cord=0, y_cord=0):
        super().__init__(x_cord, y_cord)
        self.__radius=radius

    # getters
    def radius(self):
        return self.__radius

    # setters
    def set_radius(self, radius):
        self.__radius=radius

    def compute_area(self):
        return math.pi * (self.__radius ** 2)

    def compute_perimeter(self):
        return 2 * math.pi * self.__radius
        
    def get_perimeter_points(self):
        pts = []
        x, y = self.x_cord(), self.y_cord()
        r = self.radius()
        
        # 16 points along the circumference
        for i in range(16):
            angle = 2 * math.pi * (i / 16)
            pts.append((x + r * math.cos(angle), y + r * math.sin(angle)))
        return pts


class Triangle(ShapeCalculations):
    def __init__(self, a=0.0, b=0.0, c=0.0, x_cord=0.0, y_cord=0.0):
        super().__init__(x_cord, y_cord)
        self.__a = float(a)
        self.__b = float(b)
        self.__c = float(c)

    # getters
    def a(self):
        return self.__a

    def b(self):
        return self.__b

    def c(self):
        return self.__c

    # setters
    def set_a(self, a):
        self.__a = float(a)

    def set_b(self, b):
        self.__b = float(b)

    def set_c(self, c):
        self.__c = float(c)

    def compute_perimeter(self):
        return self.__a + self.__b + self.__c

    def compute_area(self):
        s = self.compute_perimeter() / 2
        area = s * (s - self.__a) * (s - self.__b) * (s - self.__c)
        return math.sqrt(area)
        
    def get_perimeter_points(self):
        pts = []
        x, y = self.x_cord(), self.y_cord()
        a, b, c = self.a(), self.b(), self.c()
        if a == 0: return pts
        
        # calculate 3 vertices based on side lengths
        x3_rel = (a**2 + b**2 - c**2) / (2 * a)
        y3_rel = math.sqrt(max(0, b**2 - x3_rel**2))
        
        v1 = (x, y)
        v2 = (x + a, y)
        v3 = (x + x3_rel, y + y3_rel)
        
        # (5 + 5 + 6 = 16 points)
        def interp(p1, p2, n):
            return [(p1[0] + (p2[0]-p1[0])*(i/n), p1[1] + (p2[1]-p1[1])*(i/n)) for i in range(n)]
            
        pts.extend(interp(v1, v2, 5))
        pts.extend(interp(v2, v3, 5))
        pts.extend(interp(v3, v1, 6))
        return pts

My_Rectangle = Rectangle()
My_Rectangle.set_length(10)
My_Rectangle.set_width(15)
My_Rectangle.set_x_cord(0)
My_Rectangle.set_y_cord(0)
print("--Rectangle--")
print(f"perimeter points: {My_Rectangle.get_perimeter_points()}")

My_Circle = Circle()
My_Circle.set_radius(5)
My_Circle.set_x_cord(10)
My_Circle.set_y_cord(10)
print("\n--Circle--")
print(f"perimeter points: {My_Circle.get_perimeter_points()}")

My_Triangle = Triangle()
My_Triangle.set_a(3)
My_Triangle.set_b(4)
My_Triangle.set_c(5)
My_Triangle.set_x_cord(0)
My_Triangle.set_y_cord(0)
print("\n--Triangle--")
print(f"perimeter points: {My_Triangle.get_perimeter_points()}")
```

10. Copy the `Canvas` class from lecture into a python file creating a `paint` module. Copy your classes from above into the module and implement paint functions. Implement a `CompoundShape` class. Create a simple drawing demonstrating that all of your classes are working.

```python
import math

class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for _ in range(height)]

    def set_pixel(self, row, col, char='*'):
        if 0 <= row < self.height and 0 <= col < self.width:
            self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]
    
    def clear_canvas(self):
        self.data = [[' '] * self.width for _ in range(self.height)]
    
    def v_line(self, x, y, h, **kwargs):
        for i in range(x, x + h):
            self.set_pixel(i, y, **kwargs)

    def h_line(self, x, y, w, **kwargs):
        for i in range(y, y + w):
            self.set_pixel(x, i, **kwargs)
            
    def line(self, x1, y1, x2, y2, **kwargs):
        if y2 == y1:
            self.h_line(x1, min(y1, y2), abs(y2 - y1) + 1, **kwargs)
            return
        slope = (x2 - x1) / (y2 - y1)
        step = 1 if y2 > y1 else -1
        for y in range(y1, y2 + step, step):
            x = x1 + int(slope * (y - y1))
            self.set_pixel(x, y, **kwargs)
            
    def display(self):
        print("\n".join(["".join(row) for row in self.data]))

class Shape:
    def __init__(self, x_cord=0, y_cord=0, name="", **kwargs):
        self._x_cord = int(x_cord)
        self._y_cord = int(y_cord)
        self.name = name
        self.kwargs = kwargs

    def x_cord(self): return self._x_cord
    def y_cord(self): return self._y_cord
    def set_x_cord(self, x_cord): self._x_cord = int(x_cord)
    def set_y_cord(self, y_cord): self._y_cord = int(y_cord)
    
    def _kwargs_repr(self):
        kws = ", ".join(f"{k}='{v}'" if isinstance(v, str) else f"{k}={v}" for k, v in self.kwargs.items())
        return f", {kws}" if kws else ""

    def paint(self, canvas): 
        raise NotImplementedError

class Rectangle(Shape):
    def __init__(self, length=0, width=0, x_cord=0, y_cord=0, name="", **kwargs):
        super().__init__(x_cord, y_cord, name, **kwargs)
        self._length = int(length)
        self._width = int(width)

    def length(self): return self._length
    def width(self): return self._width
    def set_length(self, length): self._length = int(length)
    def set_width(self, width): self._width = int(width)

    def paint(self, canvas):
        x, y = self.x_cord(), self.y_cord()
        w, l = self.width(), self.length()
        canvas.h_line(x, y, w + 1, **self.kwargs)
        canvas.h_line(x + l, y, w + 1, **self.kwargs)
        canvas.v_line(x, y, l + 1, **self.kwargs)
        canvas.v_line(x, y + w, l + 1, **self.kwargs)

    def __repr__(self):
        return f'Rectangle({self.length()}, {self.width()}, {self.x_cord()}, {self.y_cord()}, "{self.name}"{self._kwargs_repr()})'

class Circle(Shape):
    def __init__(self, radius=0, x_cord=0, y_cord=0, name="", **kwargs):
        super().__init__(x_cord, y_cord, name, **kwargs)
        self._radius = int(radius)

    def radius(self): return self._radius
    def set_radius(self, radius): self._radius = int(radius)

    def paint(self, canvas):
        r = self.radius()
        x, y = self.x_cord(), self.y_cord()
        circumference = int(2 * math.pi * r)
        points = max(16, circumference * 2)
        for i in range(points):
            angle = 2 * math.pi * (i / points)
            px = int(round(x + r * math.cos(angle)))
            py = int(round(y + r * math.sin(angle)))
            canvas.set_pixel(px, py, **self.kwargs)

    def __repr__(self):
        return f'Circle({self.radius()}, {self.x_cord()}, {self.y_cord()}, "{self.name}"{self._kwargs_repr()})'

class Triangle(Shape):
    def __init__(self, a=0, b=0, c=0, x_cord=0, y_cord=0, name="", **kwargs):
        super().__init__(x_cord, y_cord, name, **kwargs)
        self._a, self._b, self._c = int(a), int(b), int(c)

    def a(self): return self._a
    def b(self): return self._b
    def c(self): return self._c
    def set_a(self, a): self._a = int(a)
    def set_b(self, b): self._b = int(b)
    def set_c(self, c): self._c = int(c)

    def paint(self, canvas):
        a, b, c = self.a(), self.b(), self.c()
        if a == 0 or (a + b <= c): return # Validity check
        
        x, y = self.x_cord(), self.y_cord()
        x3_rel = (a**2 + b**2 - c**2) / (2 * a)
        y3_rel = math.sqrt(max(0, b**2 - x3_rel**2))
        
        v1 = (x, y)
        v2 = (x + a, y)
        v3 = (int(x + x3_rel), int(y + y3_rel))
        
        canvas.line(v1[0], v1[1], v2[0], v2[1], **self.kwargs)
        canvas.line(v2[0], v2[1], v3[0], v3[1], **self.kwargs)
        canvas.line(v3[0], v3[1], v1[0], v1[1], **self.kwargs)

    def __repr__(self):
        return f'Triangle({self.a()}, {self.b()}, {self.c()}, {self.x_cord()}, {self.y_cord()}, "{self.name}"{self._kwargs_repr()})'

class CompoundShape(Shape):
    def __init__(self, shapes, name="", **kwargs):
        super().__init__(0, 0, name, **kwargs)
        self.shapes = shapes

    def paint(self, canvas):
        for s in self.shapes:
            s.paint(canvas)

    def __repr__(self):
        shapes_repr = "[" + ", ".join(repr(s) for s in self.shapes) + "]"
        return f'CompoundShape({shapes_repr}, "{self.name}"{self._kwargs_repr()})'
```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python
class RasterDrawing:
    def __init__(self, initial_shapes=None):
        self.shapes = dict()
        self.shape_names = list()
        
        if initial_shapes:
            for s in initial_shapes:
                self.add_shape(s)
        
    def add_shape(self, shape):
        if shape.name == "":
            shape.name = self.assign_name()
        
        self.shapes[shape.name] = shape
        if shape.name not in self.shape_names:
            self.shape_names.append(shape.name)

    def update(self, canvas):
        canvas.clear_canvas()
        self.paint(canvas)
        
    def paint(self, canvas):
        for shape_name in self.shape_names:
            self.shapes[shape_name].paint(canvas)
            
    def assign_name(self):
        name_base = "shape"
        name = f"{name_base}_0"
        i = 1
        while name in self.shapes:
            name = f"{name_base}_{i}"
            i += 1
        return name

    def __repr__(self):
        shapes_repr = "[" + ", ".join(repr(self.shapes[name]) for name in self.shape_names) + "]"
        return f'RasterDrawing({shapes_repr})'
        
    def save(self, filename):
        with open(filename, "w") as f:
            f.write(repr(self))
```

```python
c1 = Canvas(30, 30)
drawing = RasterDrawing()

drawing.add_shape(Circle(5, 15, 15, char="o"))
drawing.add_shape(Rectangle(4, 4, 2, 2, char="+"))
drawing.paint(c1)
print("---Initial drawing---")
c1.display()

drawing.shapes['shape_1'].set_length(10)
drawing.shapes['shape_1'].set_width(10)

drawing.update(c1)
print("\n---Modified drawing---")
c1.display()
```

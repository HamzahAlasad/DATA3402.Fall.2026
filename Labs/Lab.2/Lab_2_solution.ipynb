*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here

def get_winner(board):
    n = len(board)

    # check rows
    for r in range(n):
        if len(set(board[r])) == 1 and board[r][0] != empty:
            return board[r][0]

    # check columns
    for c in range(n):
        col_vals = [board[r][c] for r in range(n)]
        if len(set(col_vals)) == 1 and col_vals[0] != empty:
            return col_vals[0]

    # check primary diagonal
    diag1 = [board[i][i] for i in range(n)]
    if len(set(diag1)) == 1 and diag1[0] != empty:
        return diag1[0]

    # check secondary diagonal
    diag2 = [board[i][n - 1 - i] for i in range(n)]
    if len(set(diag2)) == 1 and diag2[0] != empty:
        return diag2[0]

    # check for empty spaces (incomplete game)
    for r in range(n):
        if empty in board[r]:
            return -1

    # if no empty space no winner, it's a draw
    return 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here

def place_mark(board, player, location):
    r, c = location
    n = len(board)
    m = len(board[0])

    # check boundaries
    if r < 0 or r >= n or c < 0 or c >= m:
        return False

    # check if empty
    if board[r][c] == empty:
        board[r][c] = player
        return True

    return False
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
# Write your solution here

row_names = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")

def parse_location(loc_str, n, m):
    if len(loc_str) < 2:
        return None

    r_char = loc_str[0].upper()
    c_str = loc_str[1:]

    if r_char not in row_names:
        return None
    r = row_names.index(r_char)

    if not c_str.isdigit():
        return None
    c = int(c_str) - 1

    if r >= n or c < 0 or c >= m:
        return None

    return (r, c)

def apply_labeled_move(board, player, loc_str):
    n = len(board)
    m = len(board[0])
    loc = parse_location(loc_str, n, m)
    if loc is None:
        return False
    return place_mark(board, player, loc)
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here

import copy

def get_valid_moves(board):
    moves = []
    n = len(board)
    for r in range(n):
        for c in range(n):
            if board[r][c] == empty:
                moves.append((r, c))
    return moves

def heuristic_ai_move(board, player):
    if player == player_1:
        opponent = player_2
    else:
        opponent = player_1

    valid_moves = get_valid_moves(board)

    # try to find a winning move
    for move in valid_moves:
        test_board = copy.deepcopy(board)
        place_mark(test_board, player, move)
        if get_winner(test_board) == player:
            return move

    # try to block opponent's winning move
    for move in valid_moves:
        test_board = copy.deepcopy(board)
        place_mark(test_board, opponent, move)
        if get_winner(test_board) == opponent:
            return move

    # pick the first available open space
    if valid_moves:
        return valid_moves[0]
    return None
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here

def score_board(board, maximizing_player):
    winner = get_winner(board)
    if winner == maximizing_player:
        return 10
    elif winner != 0 and winner != -1:
        return -10
    return 0

def exhaustive_search(board, current_player, maximizing_player, depth):
    status = get_winner(board)

    # stop if game over or depth limit reached
    if status != -1 or depth == 0:
        return score_board(board, maximizing_player)

    moves = get_valid_moves(board)

    # evaluate children
    if current_player == maximizing_player:
        best_score = -float('inf')
        for move in moves:
            next_board = copy.deepcopy(board)
            place_mark(next_board, current_player, move)
            next_player = player_2 if current_player == player_1 else player_1
            score = exhaustive_search(next_board, next_player, maximizing_player, depth - 1)
            if score > best_score:
                best_score = score
        return best_score
    else:
        best_score = float('inf')
        for move in moves:
            next_board = copy.deepcopy(board)
            place_mark(next_board, current_player, move)
            next_player = player_2 if current_player == player_1 else player_1
            score = exhaustive_search(next_board, next_player, maximizing_player, depth - 1)
            if score < best_score:
                best_score = score
        return best_score

def exhaustive_ai_move(board, player, depth):
    best_score = -float('inf')
    best_move = None
    moves = get_valid_moves(board)

    for move in moves:
        test_board = copy.deepcopy(board)
        place_mark(test_board, player, move)
        next_player = player_2 if player == player_1 else player_1
        score = exhaustive_search(test_board, next_player, player, depth - 1)

        if score > best_score:
            best_score = score
            best_move = move

    return best_move if best_move else moves[0]
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here

def ai_vs_ai_series(size, num_games, depth_p1, depth_p2):
    p1_wins = 0
    p2_wins = 0
    draws = 0

    for _ in range(num_games):
        board = make_board(size)
        current_player = player_1

        while True:
            if current_player == player_1:
                move = exhaustive_ai_move(board, current_player, depth_p1)
            else:
                move = heuristic_ai_move(board, current_player)

            place_mark(board, current_player, move)
            status = get_winner(board)

            if status != -1:
                if status == player_1:
                    p1_wins += 1
                elif status == player_2:
                    p2_wins += 1
                else:
                    draws += 1
                break

            current_player = player_2 if current_player == player_1 else player_1

    print(f"Board: {size}x{size} | Depth P1: {depth_p1} | Games: {num_games}")
    print(f"P1 (Exhaustive) Wins: {p1_wins} | P2 (Heuristic) Wins: {p2_wins} | Draws: {draws}")
````

````python
# Test your solution here

ai_vs_ai_series(3, 10, 3, 2)
ai_vs_ai_series(4, 10, 2, 2)
ai_vs_ai_series(5, 10, 2, 2)
````
